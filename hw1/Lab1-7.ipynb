````python
for dtype in sorted(set(np.sctypeDict.values()), key=str):
    dt = np.dtype(dtype)
    if dt.kind in 'iu':
        info = np.iinfo(dt)
        print(dt, info.min, info.max)
    elif dt.kind in 'fc':
        info = np.finfo(dt)
        print(dt, info.min, info.max, '(для каждой компоненты)' if dt.kind == 'c' else '')
    elif dt.kind == 'b': print(dt, False, True)
````

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
q30, q60, q75 = df['BMI'].quantile([.3, .6, .75])
has_disease = ~df['Хронические заболевания'].str.strip().str.lower().isin(['нет', ''])
groups = {'Выше 75-го перцентиля': df['BMI'] > q75, 'От 30-го до 60-го перцентиля': df['BMI'].between(q30, q60)}
result = pd.DataFrame({name: {'count': int(mask.sum()), 'disease_share': has_disease[mask].mean()} for name, mask in groups.items()}).T
display(result)
print('Заболевания чаще выше 75-го перцентиля:', result.loc['Выше 75-го перцентиля', 'disease_share'] > result.loc['От 30-го до 60-го перцентиля', 'disease_share'])
````

9. Правда ли, что люди с высоким уровнем дохода имеют больший уровень физической активности, чем люди с низким доходом

````python
activity = pd.crosstab(df['Уровень дохода'], df['Физическая активность'], normalize='index').reindex(index=['Низкий', 'Средний', 'Высокий'], columns=['Низкий', 'Умеренный', 'Высокий'])
display(activity)
print('Высокая активность чаще при высоком доходе:', activity.loc['Высокий', 'Высокий'] > activity.loc['Низкий', 'Высокий'])
````

10. Какая доля мужчин с высокой частотой пребывания на солнце имеет высокий заработок?

````python
sun_threshold = df['Пребывание на солнце'].quantile(.75)
men_sun = df.loc[df['Пол'].eq('Мужской') & (df['Пребывание на солнце'] >= sun_threshold)]
print('Порог пребывания на солнце:', sun_threshold)
print('Число мужчин в группе:', len(men_sun))
print(f"Доля с высоким доходом: {men_sun['Уровень дохода'].eq('Высокий').mean():.2%}")
````
